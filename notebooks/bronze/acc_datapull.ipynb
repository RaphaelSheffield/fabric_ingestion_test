```python
import requests
import pandas as pd

def get_project_forms(access_token: str, project_id: str) -> pd.DataFrame:
    access_token = access_token.strip()

    headers = {
        "Authorization": f"Bearer {access_token}",
        "Content-Type": "application/json",
    }

    url = (
        f"https://developer.api.autodesk.com/construction/forms/v1/"
        f"projects/{project_id}/forms?limit=50&offset=0"
    )

    response = requests.get(url, headers=headers, timeout=60)

    if response.status_code != 200:
        raise RuntimeError(
            f"Request failed with status {response.status_code}: {response.text}"
        )

    data = response.json()
    forms = data.get("data", [])

    return pd.json_normalize(forms)
```

```python
test_project_id = "b2af28b4-2a18-453e-8b28-1ed8f6a7fd6c"
df_forms = get_project_forms(access_token, test_project_id)
display(df_forms)
```
